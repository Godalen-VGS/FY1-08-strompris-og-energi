## Energi og effekt

| Størrelse | Formel | Enhet |
| --- | --- | --- |
| Energi fra effekt | $E = P\,t$ | J (W·s), eller kWh |
| 1 kWh | $1000\ \text{W} \cdot 3600\ \text{s}$ | $3{,}6 \cdot 10^6$ J |
| Varme for å varme vann | $Q = m\,c\,\Delta T$ | J |
| Spesifikk varmekapasitet for vann | $c = 4186$ J/(kg·K) | |

En vannkoker på 2000 W som står på i 3 minutter bruker $2\ \text{kW} \cdot 0{,}05\ \text{h} = 0{,}1$ kWh.

```python
import requests
from datetime import date

def hent_priser(dag, sone="NO2"):
    adresse = f"https://www.hvakosterstrommen.no/api/v1/prices/{dag.year}/{dag.month:02d}-{dag.day:02d}_{sone}.json"
    return requests.get(adresse, timeout=10).json()


data = hent_priser(date.today())
print(len(data), "prisperioder i dag")
print(data[0])
```
